```python
def basic_calculator(a, b):
    return {
        "sum": a + b,
        "difference": a - b,
        "product": a * b
    }
```

```python
def bytes_to_kb_mb(num_bytes):
    kb = round(num_bytes / 1024, 2)
    mb = round(num_bytes / (1024 * 1024), 2)
    return (kb, mb)
```

```python
def classify_number(n):
    if n % 2 == 0:
        return "even"
    else:
        return "odd"
```

```python
def build_user_record(name, age):
    return {
        "name": name,
        "age": age,
        "is_adult": age >= 18
    }
```
